```python
def load_jsonl_safely(file_path, subject):
    
    valid_records = []
    error_records = []

    with open(file_path, "r", encoding="utf-8") as file:

        for line_number, line in enumerate(file, start=1):

            line = line.strip()

            if not line:
                continue

            try:
                record = json.loads(line)

                record["subject"] = subject
                record["source_file"] = file_path.name
                record["source_line"] = line_number

                valid_records.append(record)

            except json.JSONDecodeError as error:

                error_records.append({
                    "subject": subject,
                    "source_file": file_path.name,
                    "source_line": line_number,
                    "error": str(error),
                    "raw_text": line
                })

    return valid_records, error_records
```

```python
all_records = []
all_errors = []

for file_path in jsonl_files:

    file_name = file_path.name.lower()

    if "biology" in file_name:
        subject = "Biology"

    elif "chemistry" in file_name:
        subject = "Chemistry"

    elif "physics" in file_name:
        subject = "Physics"

    else:
        subject = "Unknown"

    records, errors = load_jsonl_safely(
        file_path=file_path,
        subject=subject
    )

    all_records.extend(records)
    all_errors.extend(errors)
```

```python
def build_clean_dataframe(df_input):

    df_clean = df_input.copy()

    # -----------------------------------
    # 1. Standardize Question schema
    # -----------------------------------
    if "question" in df_clean.columns:
        df_clean["Question"] = df_clean["Question"].fillna(
            df_clean["question"]
        )

    # -----------------------------------
    # 2. Standardize Hints schema
    # -----------------------------------
    if "hints" in df_clean.columns:
        df_clean["Hints"] = df_clean["Hints"].fillna(
            df_clean["hints"]
        )

    # -----------------------------------
    # 3. Remove duplicate schema columns
    # -----------------------------------
    columns_to_drop = [
        column
        for column in ["question", "hints"]
        if column in df_clean.columns
    ]

    df_clean = df_clean.drop(
        columns=columns_to_drop
    )

    # -----------------------------------
    # 4. Normalize ExactAnswer
    # -----------------------------------
    df_clean["ExactAnswer"] = df_clean["ExactAnswer"].apply(
        lambda value:
            value if isinstance(value, list)
            else [value] if isinstance(value, str)
            else value
    )

    # -----------------------------------
    # 5. Remove empty Sources metadata
    # -----------------------------------
    if "Sources" in df_clean.columns:
        df_clean = df_clean.drop(
            columns=["Sources"]
        )

    # -----------------------------------
    # 6. Preserve original TopicTags
    # -----------------------------------
    df_clean["TopicTags_original"] = df_clean[
        "TopicTags"
    ].apply(
        lambda value:
            value.copy()
            if isinstance(value, list)
            else value
    )

    # -----------------------------------
    # 7. Repair known Chemistry tag shift
    # -----------------------------------
    chemistry_mask = (
        df_clean["source_file"]
        == "SSC_Chemistry_Dataset.jsonl"
    )

    for line_number in range(2555, 2584):

        previous_mask = (
            chemistry_mask
            & (df_clean["source_line"] == line_number - 1)
        )

        current_mask = (
            chemistry_mask
            & (df_clean["source_line"] == line_number)
        )

        previous_tags = df_clean.loc[
            previous_mask,
            "TopicTags_original"
        ].iloc[0]

        current_index = df_clean.index[
            current_mask
        ][0]

        df_clean.at[
            current_index,
            "TopicTags"
        ] = previous_tags.copy()

    # -----------------------------------
    # 8. Impute Chemistry line 2554
    #    using the near-duplicate line 2588
    # -----------------------------------
    target_mask = (
        chemistry_mask
        & (df_clean["source_line"] == 2554)
    )

    reference_mask = (
        chemistry_mask
        & (df_clean["source_line"] == 2588)
    )

    reference_tags = df_clean.loc[
        reference_mask,
        "TopicTags_original"
    ].iloc[0]

    target_index = df_clean.index[
        target_mask
    ][0]

    df_clean.at[
        target_index,
        "TopicTags"
    ] = reference_tags.copy()

    return df_clean
```
